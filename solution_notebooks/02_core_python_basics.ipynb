# Chapter 2: Core Python Basics

Let's look at the different kinds of values Python can work with, and how to combine them.

## 1. Whole Numbers and Decimals

Python has whole numbers (like `10`) and numbers with a decimal point (like `3.5`). Give your variables short, clear, lowercase names so they're easy to read.

**Task:** Store a whole number and a decimal number in two variables, then print both.

```python
whole_number = 10
decimal_number = 3.5

print(whole_number)
print(decimal_number)
```

## 2. Text and True/False Values

Text is called a string. A value that is either `True` or `False` is called a boolean.

**Task:** Store a piece of text and a `True`/`False` value in two variables, then print both.

```python
city = "London"
is_raining = True

print(city)
print(is_raining)
```

## 3. A Special "Nothing" Value

Python has a special value called `None`, which simply means "no value here".

**Task:** Store `None` in a variable called `result`, then print it.

```python
result = None
print(result)
```

## 4. Checking a Value's Type

The `type()` function tells you what kind of value something is.

**Task:** Print the type of `whole_number`, `city`, and `is_raining`.

```python
print(type(whole_number))
print(type(city))
print(type(is_raining))
```

## 5. Converting Between Types

Sometimes you need to change a value from one type to another - for example, turning text into a number.

**Task:** Convert the text `"42"` into a whole number, then print it.

```python
text_number = "42"
converted_number = int(text_number)

print(converted_number)

print(type(converted_number))
```

## 6. Arithmetic Operators

Arithmetic operators do math: `+` add, `-` subtract, `*` multiply, `/` divide.

**Task:** Store two numbers, then print their sum, difference, product, and division result.

```python
a = 10
b = 3

print(a + b)
print(a - b)
print(a * b)
print(a / b)
```

## 7. Comparison Operators

Comparison operators compare two values and give back `True` or `False`: `==` equal to, `!=` not equal to, `>` greater than, `<` less than.

**Task:** Compare `a` and `b` using `==`, `!=`, and `>`, then print each result.

```python
print(a == b)
print(a != b)
print(a > b)
```

## 8. Logical Operators

Logical operators combine `True`/`False` values: `and`, `or`, `not`.

**Task:** Store two `True`/`False` values, then combine them with `and`, `or`, and `not`.

```python
is_sunny = True
is_warm = False

print(is_sunny and is_warm)
print(is_sunny or is_warm)
print(not is_sunny)
```

## 9. Assignment Shortcuts

You can update a variable in one step with shortcuts like `+=`, which adds to the variable and saves the result back into it.

**Task:** Start with a score of `0`, add `5` to it using `+=`, print it, then add `5` again and print it. Also try `-=` for inverse of the same

```python
score = 0

score += 5
print(score)

score += 5
print(score)
```

## 10. Same Value vs Same Thing

`==` checks if two values are equal. `is` checks if two variables are the exact same thing behind the scenes - even when their values match, they aren't always the same thing.

**Task:** Create two separate variables that both hold the number `1000`, then compare them with `==` and with `is`.

```python
first_number = int("1000")
second_number = int("1000")

print(first_number == second_number)
print(first_number is second_number)
```

## 11. Putting It Together

Let's use several of these ideas in one small script.

**Task:** Store a price and a quantity, calculate the total cost, and print a message that says whether the total is over `50`.

```python
price = 12.5
quantity = 5

total_cost = price * quantity
is_over_50 = total_cost > 50

print("Total cost:", total_cost)
print("Is it over 50?", is_over_50)
```

## Recap

- Python's basic types include whole numbers, decimals, text, `True`/`False`, and `None`.
- `type()` tells you what kind of value something is; functions like `int()` convert between types.
- Arithmetic operators do math; comparison operators give back `True`/`False`; logical operators combine `True`/`False` values.
- Shortcuts like `+=` update a variable in one step.
- `==` compares values; `is` checks whether two variables are the exact same thing.
