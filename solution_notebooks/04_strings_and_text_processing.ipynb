# Chapter 4: Strings and Text Processing

Let's learn how to work with text in Python - reading it, changing it, and pulling out the pieces we need.

## 1. Storing Text in a String

A string is a piece of text stored in a variable, written between quotes.

**Task:** Store the text `"Hello, Narender!"` in a variable and print it.

```python
greeting = "Hello, Narender!"
print(greeting)
```

## 2. Storing Text Across Multiple Lines

Triple quotes (`"""`) let you store text that spans more than one line.

**Task:** Store a two-line message using triple quotes, then print it.

```python
message = """Hello, Narender!
Welcome to Python."""
print(message)
```

## 3. Finding the Length of a String with len()

`len()` tells you how many characters are in a string, including spaces.

**Task:** Store the word `"Python"` in a variable, then print how many characters it has.

```python
word = "Python"
print(len(word))
```

## 4. Getting a Single Character with Indexing

Each character in a string has a position number, called an index, starting from `0`.

**Task:** Store the word `"Python"` in a variable, then print its first character.

```python
word = "Python"
print(word[0])
```

## 5. Counting from the End with Negative Indexing

Negative index numbers count backwards from the end of a string, starting at `-1` for the last character.

**Task:** Store the word `"Python"` in a variable, then print its last character using a negative index.

```python
word = "Python"
print(word[-1])
```

## 6. Getting Part of a String with Slicing

Slicing lets you pull out a piece of a string by giving a start and stop position.

**Task:** Store the word `"Python"` in a variable, then print just the first three letters.

```python
word = "Python"
print(word[0:3])
```

## 7. Choosing a Step Size While Slicing

A third number in a slice lets you skip characters, just like it does with `range()`.

**Task:** Store the word `"Python"` in a variable, then print every second letter.

```python
word = "Python"
print(word[0:6:2])
```

## 8. Changing Letter Case with upper() and lower()

String methods are built-in actions you can perform on text, written after a dot.

**Task:** Store the word `"Python"` in a variable, then print it in uppercase and in lowercase.

```python
word = "Python"
print(word.upper())
print(word.lower())
```

## 9. Removing Extra Spaces with strip()

`strip()` removes extra spaces from the beginning and end of a string.

**Task:** Store the text `"  Hello  "` with extra spaces around it, then print it after removing those spaces.

```python
text = "  Hello  "
print(text.strip())
```

## 10. Finding Text Inside a String with find()

`find()` tells you the index where a piece of text starts inside a string, or `-1` if it isn't there.

**Task:** Store the sentence `"I like cats"` in a variable, then print the index where the word `"cats"` starts.

```python
sentence = "I like cats"
print(sentence.find("cats"))
```

## 11. Replacing Text with replace()

`replace()` swaps every occurrence of one piece of text for another.

**Task:** Store the sentence `"I like cats"` in a variable, then print it with `"cats"` replaced by `"dogs"`.

```python
sentence = "I like cats"
print(sentence.replace("cats", "dogs"))
```

## 12. Joining Two Strings with Concatenation

Concatenation means joining strings together using the `+` symbol. This always creates a brand new string, since text in Python can't be changed in place.

**Task:** Store a first name and a last name in two variables, then print them joined together with a space in between.

```python
first_name = "Narender"
last_name = "Singh"
print(first_name + " " + last_name)
```

## 13. Building Text with f-strings

An f-string lets you insert variables directly inside a piece of text by placing an `f` before the quotes.

**Task:** Store a name and an age in two variables, then use an f-string to print a sentence introducing them.

```python
name = "Narender"
age = 30
print(f"{name} is {age} years old.")
```

## 14. Splitting Text into a List with split()

`split()` breaks a string into a list of smaller pieces, using a space by default.

**Task:** Store the sentence `"I like cats and dogs"` in a variable, then print it split into a list of words.

```python
sentence = "I like cats and dogs"
print(sentence.split())
```

## 15. Joining a List into Text with join()

`join()` does the opposite of `split()` - it combines a list of pieces back into one string.

**Task:** Store a list of words, then join them into one sentence with spaces in between.

```python
words = ["I", "like", "cats", "and", "dogs"]
print(" ".join(words))
```

## 16. Putting It Together: Cleaning Messy Text

Let's combine several string tools in one practical example.

**Task:** Store the messy text `"   Narender   loves PYTHON and  Coding!!   "` in a variable, then:
1. Remove the extra spaces from the start and end with `strip()`.
2. Split it into a list of words.
3. Join the words back together with single spaces in between.
4. Make the result lowercase.
5. Replace `"coding!!"` with `"programming"`.
6. Print the cleaned text and the word count, using f-strings.

```python
raw_text = "   Narender   loves PYTHON and  Coding!!   "

# Remove extra spaces from the start and end
stripped = raw_text.strip()

# Split into a list of words
words = stripped.split()

# Join the words back together with single spaces
cleaned = " ".join(words)

# Make it lowercase
cleaned = cleaned.lower()

# Replace one word with another
cleaned = cleaned.replace("coding!!", "programming")

print(f"Cleaned text: {cleaned}")
print(f"Word count: {len(words)}")
```

## Recap

- Strings can be indexed and sliced to get individual characters or pieces of text.
- String methods like `upper()`, `lower()`, `strip()`, `find()`, and `replace()` perform common text actions.
- Strings can be joined with `+` or built with f-strings.
- `split()` breaks text into a list, and `join()` combines a list back into text.
