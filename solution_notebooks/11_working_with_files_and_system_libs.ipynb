# Chapter 11: Standard Library Essentials and File Handling

Python comes with a huge collection of ready-made tools already installed, called the standard library, including tools for reading and writing files. Let's explore some of the ones you'll reach for often.

## 1. Finding Your Current Folder

The `os` tool lets your code ask the operating system questions, like which folder it's currently running in.

**Task:** Import `os`, then print your current folder using `os.getcwd()`.

```python
import os

print(os.getcwd())
```

## 2. Listing What's Inside a Folder

`os.listdir()` gives you a list of the file and folder names inside a given folder.

**Task:** Use `os.listdir(".")` to list what's inside the current folder, and print it.

```python
items = os.listdir(".")
print(items)
```

## 3. Stopping a Program Early

`sys.exit()` stops a program immediately, so any code written after it never runs.

**Task:** Import `sys`, then print the numbers `1` through `5`, but call `sys.exit()` right after printing `3`, before printing `4` and `5`. Notice `4` and `5` never appear.

```python
import sys

print(1)
print(2)
print(3)
sys.exit()
print(4)
print(5)
```

## 4. Today's Date and Time

The `datetime` tool can tell you both today's date and the exact date and time right now.

**Task:** Import `datetime`, then print `datetime.date.today()` and `datetime.datetime.now()`.

```python
import datetime

print(datetime.date.today())
print(datetime.datetime.now())
```

## 5. Writing Text to a File

`open(...)` lets you create or open a file yourself, and `"w"` mode means write, which replaces anything already in the file. Writing `with open(...) as file:` makes Python close the file for you automatically once you're done, even if something goes wrong along the way.

**Task:** Use `with open("notes.txt", "w") as file:` to write the current time as the first line using `datetime.datetime.now()`, then write `"Milk\n"` as a second line, using `.write()`. Print a confirmation message afterward.

```python
with open("notes.txt", "w") as file:
    file.write(f"{datetime.datetime.now()}\n")
    file.write("Milk\n")

print("Notes saved.")
```

## 6. Reading a File's Full Contents

Opening a file with `"r"` mode (the default) and calling `.read()` gives back everything in the file as one piece of text.

**Task:** Open `notes.txt` and print its full contents using `.read()`.

```python
with open("notes.txt") as file:
    contents = file.read()

print(contents)
```

## 7. Reading a File Line by Line

`.readlines()` gives back a list of the file's lines, which is handy for going through them one at a time.

**Task:** Open `notes.txt`, read its lines with `.readlines()`, then loop over the list and print each line with `.strip()` to drop the extra newline.

```python
with open("notes.txt") as file:
    lines = file.readlines()

for line in lines:
    print(line.strip())
```

## 8. Turning Data Into JSON Text

The `json` tool can turn Python data, like a dictionary, into text that follows a standard format, without needing a file at all — handy any time you need to send or store data as plain text. It works with nested dictionaries too.

**Task:** Create a dictionary with `"name"`, `"age"`, and a nested `"address"` dictionary holding `"city"` and `"zip_code"`. Convert it to text with `json.dumps()`, and print the text along with its type.

```python
import json

person = {
    "name": "Sam",
    "age": 25,
    "address": {"city": "Denver", "zip_code": "80202"},
}

person_text = json.dumps(person)
print(person_text)
print(type(person_text))
```

## 9. Turning JSON Text Back Into Data

`json.loads()` does the opposite of `json.dumps()` — it turns JSON text back into real Python data, like a dictionary.

**Task:** Convert `person_text` back into a dictionary with `json.loads()`, and print the result along with its type.

```python
person_again = json.loads(person_text)
print(person_again)
print(type(person_again))
```

## 10. Saving JSON to a File

`json.dump()` (no `s`) saves Python data straight into a file, combining what you just learned with writing files.

**Task:** Use `json.dump()` to save `person` into a file named `person.json`.

```python
with open("person.json", "w") as file:
    json.dump(person, file)

print("Saved person.json")
```

## 11. Loading JSON From a File

`json.load()` (no `s`) reads a JSON file back into Python data, combining what you just learned with reading files.

**Task:** Open `person.json` and read it back with `json.load()`, then print the result.

```python
with open("person.json") as file:
    loaded_person = json.load(file)

print(loaded_person)
```

## 12. Putting It Together

Let's combine `os`, `datetime`, and `json` the way a small real script might: checking a file and saving a little report about it. Real scripts also need to handle things going wrong, so this one uses a `try`/`except` block to deal with a missing file.

**Task:** Use `os.path.exists()` to check whether `notes.txt` exists. Build a dictionary with the file name, whether it exists, and today's date. Save it to `report.json` with `json.dump()`. Then, inside a `try` block, read `report.json` back with `json.load()` and print it, using `except FileNotFoundError:` to print a message instead if the file can't be found.

```python
report_file = "notes.txt"

report = {
    "file_name": report_file,
    "exists": os.path.exists(report_file),
    "checked_on": str(datetime.date.today()),
}

with open("report.json", "w") as file:
    json.dump(report, file)

try:
    with open("report.json") as file:
        saved_report = json.load(file)
    print(saved_report)
except FileNotFoundError:
    print("report.json not found.")
```

## Recap

- `os` lets your code ask questions about the operating system, like the current folder (`os.getcwd()`), a folder's contents (`os.listdir()`), and whether a file exists (`os.path.exists()`).
- `sys.exit()` stops a program early, and any code after it never runs.
- `datetime.date.today()` and `datetime.datetime.now()` give you today's date and the current date and time.
- `open()` with `with` reads and writes text files and closes them automatically, and `.write()`, `.read()`, and `.readlines()` cover the common cases.
- `json.dumps()`/`json.loads()` convert Python data to and from JSON text directly, and `json.dump()`/`json.load()` do the same straight to and from a file.
- A `try`/`except` block lets your code handle problems, like a missing file, without crashing.
