# Chapter 5: Lists, Tuples, Dictionaries, and Sets

Let's learn how Python stores collections of items - in lists, tuples, dictionaries, and sets.

![if, elif, else overview](../images/list_dict_set.png)

## 1. Storing Multiple Items in a List

A list stores several items together in one variable, written inside square brackets.

**Task:** Store three fruit names in a list and print it.

```python
fruits = ["apple", "banana", "cherry"]
print(fruits)
```

## 2. Finding the Length of a List with len()

`len()` tells you how many items are in a list.

**Task:** Store three fruit names in a list, then print how many items it has.

```python
fruits = ["apple", "banana", "cherry"]
print(len(fruits))
```

## 3. Getting a Single Item with Indexing

Each item in a list has a position number, called an index, starting from `0`.

**Task:** Store three fruit names in a list, then print the first one.

```python
fruits = ["apple", "banana", "cherry"]
print(fruits[0])
```

## 4. Counting from the End with Negative Indexing

Negative index numbers count backwards from the end of a list, starting at `-1` for the last item.

**Task:** Store three fruit names in a list, then print the last one using a negative index.

```python
fruits = ["apple", "banana", "cherry"]
print(fruits[-1])
```

## 5. Getting Part of a List with Slicing

Slicing lets you pull out a piece of a list by giving a start and stop position.

**Task:** Store four fruit names in a list, then print just the first two.

```python
fruits = ["apple", "banana", "cherry", "date"]
print(fruits[0:2])
```

## 6. Adding an Item to the End with append()

`append()` adds one new item to the end of a list.

**Task:** Store three fruit names in a list, then add `"date"` to the end and print the list.

```python
fruits = ["apple", "banana", "cherry"]
fruits.append("date")
print(fruits)
```

## 7. Adding Several Items with extend()

`extend()` adds every item from another list onto the end of a list.

**Task:** Store a list of fruits and a second list of more fruits, then extend the first list with the second and print it.

```python
fruits = ["apple", "banana"]
more_fruits = ["cherry", "date"]
fruits.extend(more_fruits)
print(fruits)
```

## 8. Adding an Item at a Specific Position with insert()

`insert()` adds a new item at a specific position, moving the other items along.

**Task:** Store three fruit names in a list, then insert `"apricot"` at position `1` and print the list.

```python
fruits = ["apple", "banana", "cherry"]
fruits.insert(1, "apricot")
print(fruits)
```

## 9. Removing the Last Item with pop()

`pop()` removes and returns the last item in a list.

**Task:** Store four fruit names in a list, then remove the last one with `pop()` and print both the removed item and the remaining list.

```python
fruits = ["apple", "banana", "cherry", "date"]
removed = fruits.pop()
print(removed)
print(fruits)
```

## 10. Removing an Item by Value with remove()

`remove()` deletes the first item that matches a given value.

**Task:** Store three fruit names in a list, then remove `"banana"` and print the list.

```python
fruits = ["apple", "banana", "cherry"]
fruits.remove("banana")
print(fruits)
```

## 11. Putting a List in Order with sort()

`sort()` rearranges the items in a list into order, changing the original list.

**Task:** Store three fruit names in a list out of order, then sort it and print it.

```python
fruits = ["cherry", "apple", "banana"]
fruits.sort()
print(fruits)
```

## 12. Getting a Sorted Copy with sorted()

`sorted()` gives you a new sorted list without changing the original one.

**Task:** Store three numbers out of order in a list, then print a sorted copy and the original list.

```python
numbers = [3, 1, 2]
print(sorted(numbers))
print(numbers)
```

## 13. Reversing a List with reverse()

`reverse()` flips the order of the items in a list.

**Task:** Store three numbers in a list, then reverse it and print it.

```python
numbers = [1, 2, 3]
numbers.reverse()
print(numbers)
```

## 14. Building a List Quickly with a List Comprehension

A list comprehension builds a new list in one line by repeating an expression for each item in another list.

**Task:** Store the numbers 1 to 5 in a list, then use a list comprehension to build a new list with each number doubled.

```python
numbers = [1, 2, 3, 4, 5]
doubled = [number * 2 for number in numbers]
print(doubled)
```

## 15. Getting Both Position and Value with enumerate()

`enumerate()` gives you both the position and the value while looping through a list.

**Task:** Store three fruit names in a list, then use `enumerate()` to print each one with its position.

```python
fruits = ["apple", "banana", "cherry"]
for index, fruit in enumerate(fruits):
    print(index, fruit)
```

## 16. Pairing Up Two Lists with zip()

`zip()` pairs up items from two lists so you can loop through them together.

**Task:** Store a list of names and a list of ages, then use `zip()` to print each name with its matching age.

```python
names = ["Narender", "Asha", "Ravi"]
ages = [30, 25, 40]
for name, age in zip(names, ages):
    print(name, age)
```

## 17. Storing Fixed Items in a Tuple

A tuple stores several items together, just like a list, but is written with round brackets.

**Task:** Store three colors in a tuple and print it.

```python
colors = ("red", "green", "blue")
print(colors)
```

## 18. Tuples Cannot Be Changed

Tuples are immutable, which means their items can't be changed after they're created. Making a change means creating a new tuple instead.

**Task:** Store three colors in a tuple, then create a new tuple that adds `"yellow"` to it and print the new tuple.

```python
colors = ("red", "green", "blue")
more_colors = colors + ("yellow",)
print(more_colors)
```

## 19. Storing Key-Value Pairs in a Dictionary

A dictionary stores information as key-value pairs, written inside curly brackets.

**Task:** Store a person's name and age in a dictionary and print it.

```python
person = {"name": "Narender", "age": 30}
print(person)
```

## 20. Getting a Value by Its Key

You can look up a value in a dictionary using its key inside square brackets.

**Task:** Store a person's name and age in a dictionary, then print just the name.

```python
person = {"name": "Narender", "age": 30}
print(person["name"])
```

## 21. Using get() to Avoid Errors

`get()` looks up a value by its key, but returns `None` instead of an error if the key doesn't exist.

**Task:** Store a person's name and age in a dictionary, then use `get()` to look up a `"city"` key that isn't there.

```python
person = {"name": "Narender", "age": 30}
print(person.get("city"))
```

## 22. Adding or Changing a Value in a Dictionary

Assigning to a key adds it if it's new, or updates it if it already exists.

**Task:** Store a person's name and age in a dictionary, then add a `"city"` key and print the dictionary.

```python
person = {"name": "Narender", "age": 30}
person["city"] = "Delhi"
print(person)
```

## 23. Listing All Keys with keys()

`keys()` gives you every key stored in a dictionary.

**Task:** Store a person's name and age in a dictionary, then print all of its keys.

```python
person = {"name": "Narender", "age": 30}
print(person.keys())
```

## 24. Listing All Values with values()

`values()` gives you every value stored in a dictionary.

**Task:** Store a person's name and age in a dictionary, then print all of its values.

```python
person = {"name": "Narender", "age": 30}
print(person.values())
```

## 25. Listing All Pairs with items()

`items()` gives you every key and value together, as pairs.

**Task:** Store a person's name and age in a dictionary, then loop through it with `items()` and print each key with its value.

```python
person = {"name": "Narender", "age": 30}
for key, value in person.items():
    print(key, value)
```

## 26. Nesting a Dictionary Inside Another

A dictionary can store another dictionary as one of its values, which is useful for grouping related information.

**Task:** Store a person's name and age, plus a nested dictionary for their address with a city and a zip code, then print the city.

```python
person = {
    "name": "Narender",
    "age": 30,
    "address": {"city": "Delhi", "zip_code": "110001"}
}
print(person["address"]["city"])
```

## 27. Removing Duplicates with a Set

A set stores items with no duplicates and no fixed order, written inside curly brackets.

**Task:** Store a list of fruits with some repeats, then turn it into a set to remove the duplicates. Print the result sorted, since sets don't keep a fixed order.

```python
fruits = ["apple", "banana", "apple", "cherry", "banana"]
unique_fruits = set(fruits)
print(sorted(unique_fruits))
```

## 28. Combining Sets with Union

The `|` symbol combines two sets into one, keeping every unique item from both.

**Task:** Store two sets of numbers, then print their union sorted.

```python
set_a = {1, 2, 3}
set_b = {3, 4, 5}
print(sorted(set_a | set_b))
```

## 29. Finding Common Items with Intersection

The `&` symbol finds the items that appear in both sets.

**Task:** Store two sets of numbers, then print the numbers that appear in both, sorted.

```python
set_a = {1, 2, 3}
set_b = {2, 3, 4}
print(sorted(set_a & set_b))
```

## 30. Finding the Difference Between Sets

The `-` symbol finds the items that are in one set but not the other.

**Task:** Store two sets of numbers, then print the numbers that are only in the first set, sorted.

```python
set_a = {1, 2, 3}
set_b = {2, 3, 4}
print(sorted(set_a - set_b))
```

## 31. Putting It Together: Building Records from Lists

Let's combine lists, dictionaries, sets, and pairing tools in one practical example.

**Task:**
1. Store a list of names and a matching list of countries, with some countries repeated.
2. Use `zip()` to build a list of dictionaries, one per person, each with a `"name"` and a `"country"` key.
3. Pull the country out of every record into a list, then use a set to find the unique countries, sorted.
4. Use `enumerate()` to print each person's record with a number in front of it.

```python
names = ["Narender", "Asha", "Ravi"]
countries = ["India", "USA", "India"]

# Build a list of dictionaries by pairing names with countries
records = [{"name": name, "country": country} for name, country in zip(names, countries)]

# Pull the country out of every record, then find the unique ones
countries_in_records = [record["country"] for record in records]
unique_countries = sorted(set(countries_in_records))

# Print each record with its position number
for index, record in enumerate(records):
    print(index, record)

print("Unique countries:", unique_countries)
```

## Recap

- Lists store ordered, changeable collections of items - indexing, slicing, and methods like `append()`, `pop()`, and `sort()` all work with them.
- List comprehensions build lists in one line, and `enumerate()`/`zip()` help loop through one or more lists together.
- Tuples are like lists but immutable - their items can't be changed after they're created.
- Dictionaries store key-value pairs, with `get()`, `keys()`, `values()`, and `items()` for working with them, and can be nested inside each other.
- Sets store unique, unordered items, and support union (`|`), intersection (`&`), and difference (`-`).
