# Chapter 7: Object-Oriented Programming - Classes and Inheritance

Let's learn how to group related information and actions together using classes, and how new classes can build on existing ones.

## 1. Defining a Class

A class is a blueprint for creating objects that group related information and actions together. You create one with the word `class`.

**Task:** Define an empty class called `Vehicle`.

```python
class Vehicle:
    pass
```

## 2. Creating an Object from a Class

Once a class is defined, you can create an object from it - this is called making an instance of the class.

**Task:** Create an object called `my_vehicle` from the `Vehicle` class, then print it.

```python
class Vehicle:
    pass

my_vehicle = Vehicle()
print(my_vehicle)
```

## 3. Adding Information with __init__

The `__init__` method runs automatically when a new object is created, and is used to set up its starting information.

**Task:** Define a `Vehicle` class with an `__init__` method that takes a `color` and stores it on the object as `self.color`. Create a vehicle and print its color.

```python
class Vehicle:
    def __init__(self, color):
        self.color = color

my_vehicle = Vehicle("Red")
print(my_vehicle.color)
```

## 4. Storing More Than One Piece of Information

`__init__` can take several pieces of information and store each one on the object.

**Task:** Add an `age` to the `Vehicle` class alongside `color`, then create a vehicle and print both.

```python
class Vehicle:
    def __init__(self, color, age):
        self.color = color
        self.age = age

my_vehicle = Vehicle("Red", 3)
print(my_vehicle.color, my_vehicle.age)
```

## 5. Creating Multiple Objects from the Same Class

Each object made from a class keeps its own separate information, even though they share the same blueprint.

**Task:** Create two vehicles with different colors and ages, then print both.

```python
class Vehicle:
    def __init__(self, color, age):
        self.color = color
        self.age = age

vehicle1 = Vehicle("Red", 3)
vehicle2 = Vehicle("Blue", 5)
print(vehicle1.color, vehicle1.age)
print(vehicle2.color, vehicle2.age)
```

## 6. Adding a Method

A method is a function defined inside a class that describes something the object can do.

**Task:** Add a `honk` method to the `Vehicle` class that prints `"Beep!"`. Create a vehicle and call the method.

```python
class Vehicle:
    def __init__(self, color, age):
        self.color = color
        self.age = age

    def honk(self):
        print("Beep!")

my_vehicle = Vehicle("Red", 3)
my_vehicle.honk()
```

## 7. Using self to Reach the Object's Own Information

Inside a method, `self` refers to the specific object the method was called on, so `self.color` reaches that object's own color.

**Task:** Add a `describe` method to the `Vehicle` class that prints a sentence using `self.color` and `self.age`.

```python
class Vehicle:
    def __init__(self, color, age):
        self.color = color
        self.age = age

    def describe(self):
        print("A " + self.color + " vehicle is " + str(self.age) + " years old.")

my_vehicle = Vehicle("Red", 3)
my_vehicle.describe()
```

## 8. Changing an Object's Information After Creation

An object's stored information can be changed at any time by assigning a new value to it.

**Task:** Create a vehicle, print its age, change the age, then print it again.

```python
class Vehicle:
    def __init__(self, color, age):
        self.color = color
        self.age = age

my_vehicle = Vehicle("Red", 3)
print(my_vehicle.age)
my_vehicle.age = 4
print(my_vehicle.age)
```

## 9. Building a New Class from an Existing One with Inheritance

Inheritance lets a new class reuse everything from an existing class, so you don't have to rewrite it. A new class that inherits from exactly one existing class is called single inheritance.

**Task:** Define a class called `Car` that inherits from `Vehicle` by writing `Vehicle` in parentheses after the class name. Create a car and call the `honk` method it inherited.

```python
class Vehicle:
    def __init__(self, color, age):
        self.color = color
        self.age = age

    def honk(self):
        print("Beep!")

class Car(Vehicle):
    pass

my_car = Car("Red", 1)
my_car.honk()
```

## 10. Adding a New Method to the Subclass

A class that inherits from another one can also have its own extra methods, on top of everything it inherited.

**Task:** Add a `drive` method to the `Car` class that prints `"Driving!"`, then call both `honk` and `drive` on a car.

```python
class Vehicle:
    def __init__(self, color, age):
        self.color = color
        self.age = age

    def honk(self):
        print("Beep!")

class Car(Vehicle):
    def drive(self):
        print("Driving!")

my_car = Car("Red", 1)
my_car.honk()
my_car.drive()
```

## 11. Overriding a Method in the Subclass

A subclass can replace a method it inherited by defining a method with the same name - this is called overriding.

**Task:** In the `Car` class, override `honk` so it prints `"Beep beep!"` instead.

```python
class Vehicle:
    def __init__(self, color, age):
        self.color = color
        self.age = age

    def honk(self):
        print("Beep!")

class Car(Vehicle):
    def honk(self):
        print("Beep beep!")

my_car = Car("Red", 1)
my_car.honk()
```

## 12. Calling the Parent Class's Version with super()

`super()` lets a subclass call the original version of a method from the class it inherited from, even after overriding it.

**Task:** In the `Car` class, override `honk` so it first calls the parent's `honk` with `super()`, then prints an extra line.

```python
class Vehicle:
    def __init__(self, color, age):
        self.color = color
        self.age = age

    def honk(self):
        print("Beep!")

class Car(Vehicle):
    def honk(self):
        super().honk()
        print("(a friendly car horn)")

my_car = Car("Red", 1)
my_car.honk()
```

## 13. Giving the Subclass Its Own __init__ with super()

A subclass can define its own `__init__`, use `super().__init__()` to set up everything the parent class needs, and then add anything extra.

**Task:** Give `Car` its own `__init__` that takes `color`, `age`, and `doors`, calls `super().__init__()` for `color` and `age`, and stores `doors` itself.

```python
class Vehicle:
    def __init__(self, color, age):
        self.color = color
        self.age = age

class Car(Vehicle):
    def __init__(self, color, age, doors):
        super().__init__(color, age)
        self.doors = doors

my_car = Car("Red", 1, 4)
print(my_car.color, my_car.age, my_car.doors)
```

## 14. Building on a Subclass - Multilevel Inheritance

A subclass can itself become the parent of another class, forming a chain - this is called multilevel inheritance.

**Task:** Define a class called `ElectricCar` that inherits from `Car`, adds a `battery_range` attribute using `super().__init__()`, and adds a `charge` method that prints `"Charging!"`. Create an electric car and call `honk`, `drive`, and `charge` on it.

```python
class Vehicle:
    def __init__(self, color, age):
        self.color = color
        self.age = age

    def honk(self):
        print("Beep!")

class Car(Vehicle):
    def __init__(self, color, age, doors):
        super().__init__(color, age)
        self.doors = doors

    def drive(self):
        print("Driving!")

class ElectricCar(Car):
    def __init__(self, color, age, doors, battery_range):
        super().__init__(color, age, doors)
        self.battery_range = battery_range

    def charge(self):
        print("Charging!")

my_electric_car = ElectricCar("Red", 1, 4, 300)
my_electric_car.honk()
my_electric_car.drive()
my_electric_car.charge()
```

## 15. Two Subclasses from One Parent - Hierarchical Inheritance

More than one class can inherit from the same parent class - this is called hierarchical inheritance.

**Task:** Define a class called `Truck` that also inherits from `Vehicle` and adds a `load_capacity` attribute using `super().__init__()`. Create a `Car` and a `Truck`, then call `honk` on both to show they both inherited it from `Vehicle`.

```python
class Vehicle:
    def __init__(self, color, age):
        self.color = color
        self.age = age

    def honk(self):
        print("Beep!")

class Car(Vehicle):
    def __init__(self, color, age, doors):
        super().__init__(color, age)
        self.doors = doors

class Truck(Vehicle):
    def __init__(self, color, age, load_capacity):
        super().__init__(color, age)
        self.load_capacity = load_capacity

my_car = Car("Red", 1, 4)
my_truck = Truck("Blue", 2, 2000)
my_car.honk()
my_truck.honk()
```

## 16. Combining Two Classes - Multiple Inheritance

A class can also inherit from more than one existing class at the same time - this is called multiple inheritance.

**Task:** Define a small `Electric` class with a `charge` method that prints `"Charging!"`. Define a `HybridCar` class that inherits from both `Vehicle` and `Electric`. Create a hybrid car and call both `honk` and `charge` on it.

```python
class Vehicle:
    def __init__(self, color, age):
        self.color = color
        self.age = age

    def honk(self):
        print("Beep!")

class Electric:
    def charge(self):
        print("Charging!")

class HybridCar(Vehicle, Electric):
    pass

my_hybrid_car = HybridCar("Green", 1)
my_hybrid_car.honk()
my_hybrid_car.charge()
```

## 17. Marking an Attribute as Protected with One Underscore

Starting an attribute's name with one underscore is a signal to other programmers that it's meant for internal use, though Python doesn't stop it from being read or changed.

**Task:** Store the vehicle's age as `self._age` instead of `self.age`, then print it using `_age`.

```python
class Vehicle:
    def __init__(self, color, age):
        self.color = color
        self._age = age

my_vehicle = Vehicle("Red", 3)
print(my_vehicle._age)
```

## 18. Marking an Attribute as Private with Two Underscores

Starting an attribute's name with two underscores makes Python rename it internally, which makes it much harder to reach from outside the class by accident.

**Task:** Store the vehicle's age as `self.__age`, add a method called `get_age` that returns it, then use the method to print the age.

```python
class Vehicle:
    def __init__(self, color, age):
        self.color = color
        self.__age = age

    def get_age(self):
        return self.__age

my_vehicle = Vehicle("Red", 3)
print(my_vehicle.get_age())
```

## 19. Printing an Object Nicely with __str__

By default, printing an object shows something unhelpful like where it's stored in memory. The `__str__` method lets you control what gets shown instead.

**Task:** Add a `__str__` method to the `Vehicle` class that returns a sentence describing the vehicle, then print a vehicle directly.

```python
class Vehicle:
    def __init__(self, color, age):
        self.color = color
        self.age = age

    def __str__(self):
        return "A " + self.color + " vehicle, age " + str(self.age)

my_vehicle = Vehicle("Red", 3)
print(my_vehicle)
```

## 20. Putting It Together: A Shape Class Hierarchy

Let's combine everything - attributes, methods, and multilevel and hierarchical inheritance - in one practical example.

**Task:**
1. Define a `Shape` class with an `__init__` that stores `color`.
2. Define a `Rectangle` class that inherits from `Shape`, adds `width` and `height` attributes using `super().__init__()`, and an `area` method that returns `width * height`.
3. Define a `Square` class that inherits from `Rectangle`, takes just `color` and `side`, and calls `super().__init__()` with `side` as both the width and height - it can reuse Rectangle's `area` method without rewriting it.
4. Create one `Rectangle` and one `Square`, then print each one's color and area.

```python
class Shape:
    def __init__(self, color):
        self.color = color

class Rectangle(Shape):
    def __init__(self, color, width, height):
        super().__init__(color)
        self.width = width
        self.height = height

    def area(self):
        return self.width * self.height

class Square(Rectangle):
    def __init__(self, color, side):
        super().__init__(color, side, side)


my_rectangle = Rectangle("blue", 3, 4)
print(my_rectangle.color, "rectangle area:", my_rectangle.area())

my_square = Square("red", 4)
print(my_square.color, "square area:", my_square.area())
```

## Recap

- A class is a blueprint for creating objects, made with `class`, and `__init__` sets up an object's starting information.
- `self` lets a method reach the specific object it was called on, and methods describe actions an object can do.
- Inheritance lets a new class reuse another class's attributes and methods, and its methods can be overridden or extended with `super()`.
- A class can inherit from one class (single inheritance), form a chain of classes (multilevel inheritance), share a parent with other classes (hierarchical inheritance), or inherit from more than one class at once (multiple inheritance) - real class hierarchies often combine these patterns.
- A single underscore signals an attribute is for internal use, a double underscore makes it harder to reach from outside, and `__str__` controls how an object looks when printed.
