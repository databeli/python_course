# Chapter 1: Your First Python Program

Let's write some real Python code, one small step at a time.

## 1. Hello, World!

`print()` shows text on the screen.

**Task:** Print the message `Hello, World!`.

```python
print("Hello, World!")
```

## 2. Printing a Number

`print()` can show numbers too, not just text.

**Task:** Print the number `7`.

```python
print(7)
```

## 3. Storing a Number and Printing It

You can save a value with a name so you can use it later. This is called a variable.

**Task:** Store the number `25` in a variable called `age`, then print it.

```python
age = 25
print(age)
```

## 4. Storing Text and Printing It

Text is stored the same way, just wrap it in quotes.

**Task:** Store the text `"Sam"` in a variable called `name`, then print it.

```python
name = "Sam"
print(name)
```

## 5. Printing Text and a Number Together

You can print several things in one line by separating them with commas.

**Task:** Print `My name is` followed by the `name` variable, and `and I am` followed by the `age` variable, and `years old.`

```python
print("My name is", name, "and I am", age, "years old.")
```

## 6. Single-Line Comments

A comment starts with `#`. Python ignores it - it's just a note for people reading the code.

**Task:** Add a comment above the line below explaining what it does, then run the cell.

```python
# This prints a greeting message
print("Hello, World!")
```

## 7. Multi-Line Comments

For longer notes, start each line with `#`.

**Task:** Write a two-line comment explaining what the code below does, then run it.

```python
# This code stores a name and an age,
# then prints them together in one sentence.
name = "Sam"
age = 25
print("My name is", name, "and I am", age, "years old.")
```

## 8. Asking the User a Question

`input()` lets the program ask a question and wait for the user to type an answer.

**Task:** Ask the user for their name, store the answer, then print a greeting.

```python
name = input("What is your name? ")
print("Hello,", name)
```

## Recap

- `print()` shows text or numbers on the screen.
- A variable stores a value so you can use it again.
- `#` starts a comment, which Python ignores.
- `input()` asks the user a question and saves their answer.
