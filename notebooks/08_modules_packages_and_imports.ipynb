# Chapter 8: Modules and Ready-Made Code

Let's learn how to reuse code by importing it from a module, and how to install and use modules other people have already written.

## 1. What Is a Module?

A module is just a Python file whose code you can reuse in another file, instead of retyping it.

**Task:** Use `%%writefile` to save a file called `greetings.py` with a function called `say_hello` that prints `"Hello!"`.

```python
%%writefile greetings.py
def say_hello():
    print("Hello!")
```

## 2. Importing a Module and Using Its Function

To reuse a module's code, bring it into your file with `import`, then reach its functions by writing the module's name, a dot, and the function name.

**Task:** Import the `greetings` module, then call `say_hello` from it.

```python
import greetings

greetings.say_hello()
```

## 3. Using a Module That's Already Built Into Python

Python comes with many modules already included, so you can use them right away without installing anything.

**Task:** Import the built-in `math` module, then print the square root of `16` using `math.sqrt(...)`.

```python
import math

print(math.sqrt(16))
```

## 4. Installing a Ready-Made Module with pip

`pip` is the tool that comes with Python for installing modules other people have written and shared, so you don't have to write everything yourself. Running the install command below will show progress messages as it downloads and sets up the module.

**Task:** Install the `python-dotenv` module by running `pip install python-dotenv` from a notebook cell, using `%pip`.

```python
%pip install python-dotenv
```

## 5. Listing Packages in a requirements.txt File

A `requirements.txt` file is a plain text list of the modules a project needs, so anyone working on it can install them all at once instead of typing each `pip install` command by hand.

**Task:** Save a `requirements.txt` file that lists two modules, `python-dotenv` and `requests`, each on its own line.

```python
%%writefile requirements.txt
python-dotenv
requests
```

## 6. Installing Every Package Listed in requirements.txt

Adding `-r` followed by a file name tells `pip` to read that file and install everything it lists, in one command.

**Task:** Install everything listed in `requirements.txt` by running `pip install -r requirements.txt`, using `%pip`.

```python
%pip install -r requirements.txt
```

## 7. Saving Settings in a .env File

A `.env` file is a plain text file that stores settings as `NAME=value` pairs, kept separate from your code.

**Task:** Save a `.env` file with two lines: `USERNAME=Sam` and `FAVORITE_COLOR=blue`.

```python
%%writefile .env
USERNAME=Sam
FAVORITE_COLOR=blue
```

## 8. Reading Values from the .env File

The `dotenv` module you installed can load a `.env` file's values, so your code can read them with `os.getenv(...)`.

**Task:** Import `load_dotenv` from `dotenv` and the built-in `os` module, call `load_dotenv()` to load the file, then print both values with `os.getenv(...)`.

```python
from dotenv import load_dotenv
import os

load_dotenv()

print(os.getenv("USERNAME"))
print(os.getenv("FAVORITE_COLOR"))
```

## Recap

- A module is a Python file whose code you can reuse elsewhere with `import`, and you reach its functions with a dot after the module's name.
- Python already comes with many built-in modules, like `math`, ready to import without installing anything.
- `pip` (run as `%pip install ...` in a notebook) adds ready-made modules that other people have written.
- A `requirements.txt` file lists a project's modules in one place, and `pip install -r requirements.txt` installs everything it lists in a single command.
- A `.env` file stores settings as `NAME=value` pairs, and the `dotenv` module can load them so `os.getenv(...)` can read them in your code.
