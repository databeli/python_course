## Recap

- Functions are defined with `def`, can take parameters, and use `return` to send back a result.
- Parameters can have default values, and arguments can be passed by position or by keyword.
- `*args` collects extra positional values into a tuple, and `**kwargs` collects extra keyword values into a dictionary.
- Variables created inside a function are local to it, but `global` lets a function change a variable from outside.
- `lambda` creates small, unnamed functions, often used with `map()`, `filter()`, and `reduce()` to transform, select, and combine items in a list.
