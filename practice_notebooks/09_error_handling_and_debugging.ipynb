## Recap

- An error stops a program unless it's handled, and `try`/`except` catches an error so the program can keep running.
- Different problems raise different built-in errors, like `ZeroDivisionError`, `ValueError`, `KeyError`, and `IndexError`, and a `try` can catch several kinds with separate `except` blocks.
- `finally` always runs, whether or not there was an error, which makes it useful for cleanup steps.
- `raise` lets your own code trigger an error on purpose, instead of letting the program crash unexpectedly.
