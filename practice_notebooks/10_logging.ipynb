## 3. The Five Levels of Notes

Every note has a level, from least to most serious. Without any setup, only `WARNING` and above show, like we just saw.

| Level | What it means |
| --- | --- |
| `DEBUG` | A small detail, useful only while building the program |
| `INFO` | A normal update about what the program is doing |
| `WARNING` | Something to watch, but not a failure |
| `ERROR` | Something failed |
| `CRITICAL` | The program can't continue |

## Recap

- `print()` works for tiny scripts, but the `logging` tool gives structured notes with levels and a choice of destination.
- The five levels, from least to most serious, are `DEBUG`, `INFO`, `WARNING`, `ERROR`, and `CRITICAL`, and without any setup only `WARNING` and above show.
