## Recap

- `requests.get()` asks another computer for data, and `.status_code` shows whether the request worked (`200` means success).
- `.json()` turns a JSON response straight into Python data, like a dictionary, that you can pick values out of with square brackets.
- The `params` dictionary sends information along with a request, like which name to look up.
- Some APIs need an API key or token, usually sent in a `headers` dictionary, to prove who's asking.
- Keys belong in their own `.env` file, not typed into code, and `load_dotenv()` plus `os.getenv()` load them in safely.
- `requests.post()` sends data-carrying requests by hand, like the Gemini chat API's address, headers, and JSON message.
- The `google-genai` SDK wraps that same kind of call into one simple line, `client.models.generate_content()`.
